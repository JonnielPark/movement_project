# 10. Biomechanical Proxy Test

This notebook tests the biomechanical proxy model on normalized pose data.

The biomechanical proxy module estimates:

- **COM (Center of Mass)** — estimated from segmental anthropometric weights
- **Moment arm** — simplified 2D joint moment arm approximation
- **Joint load distribution** — relative inter-joint load tendencies (not absolute torque)

This is a proxy model, not a full inverse dynamics solver.
All outputs are relative, body-scale-normalized indices derived from monocular 3D coordinates.
