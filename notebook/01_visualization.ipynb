```python
import sys
from pathlib import Path
import plotly.io as pio

# 프로젝트 루트
PROJECT_ROOT = Path().resolve().parent

# src import용
sys.path.append(str(PROJECT_ROOT))

# 🔥 data 경로도 같이 정의
DATA_DIR = PROJECT_ROOT / "data" / "sample"

pio.renderers.default = "browser"

print("Project root:", PROJECT_ROOT)
```

```python
from movement.io import load_pose_csv, print_data_summary
from movement.config import LANDMARKS, CONNECTIONS
from movement.utils import validate_landmark_columns
from movement.visualization import create_pose_animation

print("modules loaded")
```

```python
DATA_PATH = DATA_DIR / "mediapipe_forward_bend_sample.csv"

df = load_pose_csv(DATA_PATH)

print_data_summary(df)

df.head()
```

```python
missing_cols = validate_landmark_columns(df, LANDMARKS)

if missing_cols:
    print("Missing columns:")
    print(missing_cols[:20])
    print("Total missing:", len(missing_cols))
else:
    print("All landmark columns exist.")
```

```python
fig = create_pose_animation(
    df=df,
    landmarks=LANDMARKS,
    connections=CONNECTIONS,
    frame_duration=80,
    height=750,
    width=1000,
    padding=0.1,
    title="Forward Bend - MediaPipe 3D Pose"
)

fig.show()
```
