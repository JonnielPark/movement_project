# 12. Simulation Robustness Test

This notebook evaluates the robustness of the full pipeline against synthetic degraded data.

Synthetic data is generated by injecting controlled perturbations into normal motion sequences:

- **ROM restriction** — artificially limited joint range of motion (graded levels)
- **Visual noise** — coordinate jitter with varying sigma
- **Occlusion** — simulated landmark occlusion at varying frame ratios

Robustness checks verify that the digital biomarkers respond monotonically and
consistently as perturbation intensity increases.

```text
monotonicity  :  biomarker score degrades as perturbation increases
consistency   :  repeated runs with same parameters produce stable scores
```
