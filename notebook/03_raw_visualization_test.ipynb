# 03. Raw Visualization Test

This notebook checks whether raw pose coordinates can be visualized as a 3D skeleton animation.

Previous steps are assumed to be working:
- 00_environment_check
- 01_data_loading_test
- 02_validation_test


```python
from movement.io import load_pose_csv
from movement.config import LANDMARKS, CONNECTIONS
from movement.visualization import create_pose_animation

df = load_pose_csv("../data/sample/mediapipe_forward_bend_sample.csv")

fig = create_pose_animation(
    df=df,
    landmarks=LANDMARKS,
    connections=CONNECTIONS,
    coord_mode="raw",
    title="Raw Pose Coordinates",
    show_text=False,
)

fig.show()
```
